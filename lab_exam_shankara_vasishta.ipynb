## Clean and Standardize Metadata

### Subtask:
Iterate through each animal entry in the loaded JSON data. Apply the following transformations: 1. Standardize conservation status fields (conservation_status_, conservation, status -> conservation_status). 2. Standardize habitat fields (habitat, habitats -> habitat_type). 3. Standardize diet fields (diet, diet_type -> diet). 4. Fix typos in diet categories (e.g., 'omnivor' -> 'omnivore'). 5. Standardize habitat values (e.g., 'fresh water' -> 'freshwater'). Ensure all original relevant information is preserved under the new standardized keys.


**Reasoning**:
I need to iterate through the loaded JSON data, apply the specified standardization rules for conservation status, habitat, and diet, including typo corrections and value standardizations, and store the results in a new list.



```python
standardized_metadata = []

for animal_entry in aux_metadata_titlecase_data:
    cleaned_entry = {}

    # Handle 'animal_name' key
    cleaned_entry['animal_name'] = animal_entry.get('animal_name', 'unknown').title()

    # Handle 'conservation status' keys
    conservation_status_value = 'unknown'
    if 'conservation_status' in animal_entry:
        conservation_status_value = animal_entry['conservation_status']
    elif 'conservation' in animal_entry:
        conservation_status_value = animal_entry['conservation']
    elif 'status' in animal_entry:
        conservation_status_value = animal_entry['status']
    cleaned_entry['conservation_status'] = conservation_status_value.lower().strip()

    # Handle 'habitat' keys and standardize values
    habitat_value = 'unknown'
    if 'habitat' in animal_entry:
        habitat_value = animal_entry['habitat']
    elif 'habitats' in animal_entry:
        habitat_value = animal_entry['habitats']

    habitat_value = habitat_value.lower().strip()
    # Standardize habitat values
    habitat_mapping = {
        'fresh water': 'freshwater',
        'jungle': 'forest',
        'rainforest': 'forest',
        'desert': 'arid',
        'polar': 'arctic',
        'mountain': 'alpine',
        'ocean': 'marine',
        'river': 'freshwater',
        'lake': 'freshwater',
        'swamp': 'wetland',
        'grasslands': 'grassland'
    }
    cleaned_entry['habitat_type'] = habitat_mapping.get(habitat_value, habitat_value)

    # Handle 'diet' keys and fix typos
    diet_value = 'unknown'
    if 'diet' in animal_entry:
        diet_value = animal_entry['diet']
    elif 'diet_type' in animal_entry:
        diet_value = animal_entry['diet_type']

    diet_value = diet_value.lower().strip()
    # Correct typos in diet categories
    diet_mapping = {
        'omnivor': 'omnivore',
        'herbivore ': 'herbivore',
        'carnivore ': 'carnivore',
        'insectivore ': 'insectivore'
    }
    cleaned_entry['diet'] = diet_mapping.get(diet_value, diet_value)

    standardized_metadata.append(cleaned_entry)

print("Standardized metadata for first 5 entries:")
for entry in standardized_metadata[:5]:
    print(entry)
```

## Prepare Class Data for Merging

### Subtask:
The `class_titlecase.csv` has a comma-separated 'Animal_Names' column. This step will split these names into individual rows and merge the class information, like 'Class_Type', with each animal name. This transforms the class data into a format suitable for merging with individual animal records.


**Reasoning**:
I need to process the `class_df` by splitting the 'Animal_Names' column, exploding it into new rows, stripping whitespace, selecting the relevant columns, and renaming the animal name column for consistency, as per the instructions.



```python
import pandas as pd

# 1. Split 'Animal_Names' into lists and explode to create a new row for each animal
#    Also, strip leading/trailing whitespace from each animal name
prepared_class_df = class_df.assign(Animal_Names=class_df['Animal_Names'].str.split(', ').apply(lambda x: [name.strip() for name in x]))
prepared_class_df = prepared_class_df.explode('Animal_Names')

# 2. Select only the 'Class_Type' and the newly created 'Animal_Names' columns
prepared_class_df = prepared_class_df[['Class_Type', 'Animal_Names']]

# 3. Rename the column containing the individual animal names to 'animal_name'
prepared_class_df = prepared_class_df.rename(columns={'Animal_Names': 'animal_name'})

# 4. Display the first 5 rows and the total number of rows
print("Prepared Class DataFrame (first 5 rows):")
print(prepared_class_df.head())

print(f"\nTotal number of rows in Prepared Class DataFrame: {len(prepared_class_df)}")
```

## Create beta_eda_and_cleaning Function

### Subtask:
Consolidate all previous data processing and visualization logic into a single Python function named `beta_eda_and_cleaning`.


**Reasoning**:
The subtask requires consolidating all previous data processing and visualization logic into a single Python function named `beta_eda_and_cleaning`. I will define this function, incorporating all data loading, JSON cleaning and standardization, class data preparation, dataset merging, missing value imputation, feature engineering, and visualization generation steps as outlined in the instructions.



```python
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def beta_eda_and_cleaning():
    """
    Consolidates data loading, cleaning, merging, feature engineering, and visualization
    logic for the animal datasets.
    """

    # 1. Load Data
    zoo_df = pd.read_csv('zoo_titlecase.csv')
    class_df = pd.read_csv('class_titlecase.csv')

    with open('auxiliary_metadata_titlecase.json', 'r') as f:
        aux_metadata_titlecase_data = json.load(f)

    # 2. Clean and Standardize JSON data
    standardized_metadata = []
    for animal_entry in aux_metadata_titlecase_data:
        cleaned_entry = {}
        cleaned_entry['animal_name'] = animal_entry.get('animal_name', 'unknown').title()

        # Standardize 'conservation status' keys
        conservation_status_value = 'unknown'
        if 'conservation_status' in animal_entry:
            conservation_status_value = animal_entry['conservation_status']
        elif 'conservation' in animal_entry:
            conservation_status_value = animal_entry['conservation']
        elif 'status' in animal_entry:
            conservation_status_value = animal_entry['status']
        cleaned_entry['conservation_status'] = conservation_status_value.lower().strip()

        # Standardize 'habitat' keys and values
        habitat_value = 'unknown'
        if 'habitat' in animal_entry:
            habitat_value = animal_entry['habitat']
        elif 'habitats' in animal_entry:
            habitat_value = animal_entry['habitats']

        habitat_value = habitat_value.lower().strip()
        habitat_mapping = {
            'fresh water': 'freshwater',
            'jungle': 'forest',
            'rainforest': 'forest',
            'desert': 'arid',
            'polar': 'arctic',
            'mountain': 'alpine',
            'ocean': 'marine',
            'river': 'freshwater',
            'lake': 'freshwater',
            'swamp': 'wetland',
            'grasslands': 'grassland'
        }
        cleaned_entry['habitat_type'] = habitat_mapping.get(habitat_value, habitat_value)

        # Standardize 'diet' keys and correct typos
        diet_value = 'unknown'
        if 'diet' in animal_entry:
            diet_value = animal_entry['diet']
        elif 'diet_type' in animal_entry:
            diet_value = animal_entry['diet_type']

        diet_value = diet_value.lower().strip()
        diet_mapping = {
            'omnivor': 'omnivore',
            'herbivore ': 'herbivore',
            'carnivore ': 'carnivore',
            'insectivore ': 'insectivore'
        }
        cleaned_entry['diet'] = diet_mapping.get(diet_value, diet_value)

        standardized_metadata.append(cleaned_entry)
    aux_df = pd.DataFrame(standardized_metadata)

    # 3. Prepare Class Data for Merging
    prepared_class_df = class_df.assign(Animal_Names=class_df['Animal_Names'].str.split(', ').apply(lambda x: [name.strip() for name in x]))
    prepared_class_df = prepared_class_df.explode('Animal_Names')
    prepared_class_df = prepared_class_df[['Class_Type', 'Animal_Names']]
    prepared_class_df = prepared_class_df.rename(columns={'Animal_Names': 'animal_name'})

    # 4. Merge Datasets
    merged_df = pd.merge(zoo_df, prepared_class_df, on='animal_name', how='left')
    final_merged_df = pd.merge(merged_df, aux_df, on='animal_name', how='left')

    # 5. Impute Missing Values
    categorical_cols = final_merged_df.select_dtypes(include=['object']).columns
    numerical_cols = final_merged_df.select_dtypes(include=['int64', 'float64']).columns

    for col in categorical_cols:
        if final_merged_df[col].isnull().any():
            final_merged_df[col] = final_merged_df[col].fillna('unknown')

    for col in numerical_cols:
        if final_merged_df[col].isnull().any():
            median_value = final_merged_df[col].median()
            final_merged_df[col] = final_merged_df[col].fillna(median_value)

    # 6. Feature Engineering
    def get_habitat_risk_score(habitat_type):
        if habitat_type == 'marine':
            return 3
        elif habitat_type == 'freshwater':
            return 2
        else:
            return 1 # Includes 'terrestrial' and 'unknown'

    final_merged_df['habitat_risk_score'] = final_merged_df['habitat_type'].apply(get_habitat_risk_score)

    def get_trophic_level(diet_type):
        if diet_type == 'carnivore':
            return 3
        elif diet_type == 'omnivore':
            return 2
        else:
            return 1 # Includes 'herbivore', 'insectivore', and 'unknown'

    final_merged_df['trophic_level'] = final_merged_df['diet'].apply(get_trophic_level)

    # 7. Generate Visualizations

    # Bar plot for class counts
    plt.figure(figsize=(12, 6))
    sns.countplot(data=final_merged_df, x='Class_Type', palette='viridis', order=final_merged_df['Class_Type'].value_counts().index)
    plt.title('Distribution of Animal Classes')
    plt.xlabel('Class Type')
    plt.ylabel('Number of Animals')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Hexbin plot of 2 continuous features by class
    selected_classes_for_hexbin = ['Mammal', 'Bird', 'Fish', 'Reptile']
    for class_type in selected_classes_for_hexbin:
        class_df_hex = final_merged_df[final_merged_df['Class_Type'] == class_type]
        # Check if 'legs' and 'fins' columns have variance for hexbin plot
        if not class_df_hex.empty and class_df_hex['legs'].nunique() > 1 and class_df_hex['fins'].nunique() > 1:
            g = sns.jointplot(data=class_df_hex, x='legs', y='fins', kind='hex', cmap='Blues', height=7)
            g.fig.suptitle(f'Hexbin Plot of Legs vs Fins for {class_type} Class', y=1.02)
            plt.show()
        elif not class_df_hex.empty:
            plt.figure(figsize=(6, 6))
            sns.scatterplot(data=class_df_hex, x='legs', y='fins')
            plt.title(f'Scatter Plot of Legs vs Fins for {class_type} Class (Hexbin not suitable)')
            plt.xlabel('Legs')
            plt.ylabel('Fins')
            plt.show()
        else:
            print(f"No data for {class_type} to create hexbin plot.")

    # Swarm plot of conservation status across classes
    plt.figure(figsize=(14, 8))
    sns.swarmplot(data=final_merged_df, x='Class_Type', y='trophic_level', hue='conservation_status', palette='muted', dodge=True)
    plt.title('Trophic Level Distribution Across Classes by Conservation Status')
    plt.xlabel('Class Type')
    plt.ylabel('Trophic Level')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Conservation Status', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

    # Clustermap of all numerical features
    numerical_features_df = final_merged_df.select_dtypes(include=['int64', 'float64'])
    correlation_matrix = numerical_features_df.corr()
    sns.clustermap(correlation_matrix, cmap='coolwarm', annot=True, fmt=".2f", linewidths=.5, figsize=(12,12))
    plt.suptitle('Clustermap of Correlation Matrix for All Numerical Features', y=1.02)
    plt.show()

    return final_merged_df

print("Function `beta_eda_and_cleaning` defined successfully.")
```

```python
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def beta_eda_and_cleaning():
    """
    Consolidates data loading, cleaning, merging, feature engineering, and visualization
    logic for the animal datasets.
    """

    # 1. Load Data
    zoo_df = pd.read_csv('zoo_titlecase.csv')
    class_df = pd.read_csv('class_titlecase.csv')

    with open('auxiliary_metadata_titlecase.json', 'r') as f:
        aux_metadata_titlecase_data = json.load(f)

    # 2. Clean and Standardize JSON data
    standardized_metadata = []
    for animal_entry in aux_metadata_titlecase_data:
        cleaned_entry = {}
        cleaned_entry['animal_name'] = animal_entry.get('animal_name', 'unknown').title()

        # Standardize 'conservation status' keys
        conservation_status_value = 'unknown'
        if 'conservation_status' in animal_entry:
            conservation_status_value = animal_entry['conservation_status']
        elif 'conservation' in animal_entry:
            conservation_status_value = animal_entry['conservation']
        elif 'status' in animal_entry:
            conservation_status_value = animal_entry['status']
        cleaned_entry['conservation_status'] = conservation_status_value.lower().strip()

        # Standardize 'habitat' keys and values
        habitat_value = 'unknown'
        if 'habitat' in animal_entry:
            habitat_value = animal_entry['habitat']
        elif 'habitats' in animal_entry:
            habitat_value = animal_entry['habitats']

        habitat_value = habitat_value.lower().strip()
        habitat_mapping = {
            'fresh water': 'freshwater',
            'jungle': 'forest',
            'rainforest': 'forest',
            'desert': 'arid',
            'polar': 'arctic',
            'mountain': 'alpine',
            'ocean': 'marine',
            'river': 'freshwater',
            'lake': 'freshwater',
            'swamp': 'wetland',
            'grasslands': 'grassland'
        }
        cleaned_entry['habitat_type'] = habitat_mapping.get(habitat_value, habitat_value)

        # Standardize 'diet' keys and correct typos
        diet_value = 'unknown'
        if 'diet' in animal_entry:
            diet_value = animal_entry['diet']
        elif 'diet_type' in animal_entry:
            diet_value = animal_entry['diet_type']

        diet_value = diet_value.lower().strip()
        diet_mapping = {
            'omnivor': 'omnivore',
            'herbivore ': 'herbivore',
            'carnivore ': 'carnivore',
            'insectivore ': 'insectivore'
        }
        cleaned_entry['diet'] = diet_mapping.get(diet_value, diet_value)

        standardized_metadata.append(cleaned_entry)
    aux_df = pd.DataFrame(standardized_metadata)

    # 3. Prepare Class Data for Merging
    prepared_class_df = class_df.assign(Animal_Names=class_df['Animal_Names'].str.split(', ').apply(lambda x: [name.strip() for name in x]))
    prepared_class_df = prepared_class_df.explode('Animal_Names')
    prepared_class_df = prepared_class_df[['Class_Type', 'Animal_Names']]
    prepared_class_df = prepared_class_df.rename(columns={'Animal_Names': 'animal_name'})

    # 4. Merge Datasets
    merged_df = pd.merge(zoo_df, prepared_class_df, on='animal_name', how='left')
    final_merged_df = pd.merge(merged_df, aux_df, on='animal_name', how='left')

    # 5. Impute Missing Values
    categorical_cols = final_merged_df.select_dtypes(include=['object']).columns
    numerical_cols = final_merged_df.select_dtypes(include=['int64', 'float64']).columns

    for col in categorical_cols:
        if final_merged_df[col].isnull().any():
            final_merged_df[col] = final_merged_df[col].fillna('unknown')

    for col in numerical_cols:
        if final_merged_df[col].isnull().any():
            median_value = final_merged_df[col].median()
            final_merged_df[col] = final_merged_df[col].fillna(median_value)

    # 6. Feature Engineering
    def get_habitat_risk_score(habitat_type):
        if habitat_type == 'marine':
            return 3
        elif habitat_type == 'freshwater':
            return 2
        else:
            return 1 # Includes 'terrestrial' and 'unknown'

    final_merged_df['habitat_risk_score'] = final_merged_df['habitat_type'].apply(get_habitat_risk_score)

    def get_trophic_level(diet_type):
        if diet_type == 'carnivore':
            return 3
        elif diet_type == 'omnivore':
            return 2
        else:
            return 1 # Includes 'herbivore', 'insectivore', and 'unknown'

    final_merged_df['trophic_level'] = final_merged_df['diet'].apply(get_trophic_level)

    # 7. Generate Visualizations

    # Bar plot for class counts
    plt.figure(figsize=(12, 6))
    sns.countplot(data=final_merged_df, x='Class_Type', palette='viridis', order=final_merged_df['Class_Type'].value_counts().index)
    plt.title('Distribution of Animal Classes')
    plt.xlabel('Class Type')
    plt.ylabel('Number of Animals')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()

    # Hexbin plot of 2 continuous features by class with variance check
    selected_classes_for_hexbin = ['Mammal', 'Bird', 'Fish', 'Reptile']
    for class_type in selected_classes_for_hexbin:
        class_df_hex = final_merged_df[final_merged_df['Class_Type'] == class_type]
        # Check if 'legs' and 'fins' columns have variance for hexbin plot
        if not class_df_hex.empty and class_df_hex['legs'].nunique() > 1 and class_df_hex['fins'].nunique() > 1:
            g = sns.jointplot(data=class_df_hex, x='legs', y='fins', kind='hex', cmap='Blues', height=7)
            g.fig.suptitle(f'Hexbin Plot of Legs vs Fins for {class_type} Class', y=1.02)
            plt.show()
        elif not class_df_hex.empty:
            # Fallback to scatter plot if hexbin is not suitable due to lack of variance
            plt.figure(figsize=(6, 6))
            sns.scatterplot(data=class_df_hex, x='legs', y='fins')
            plt.title(f'Scatter Plot of Legs vs Fins for {class_type} Class (Hexbin not suitable)')
            plt.xlabel('Legs')
            plt.ylabel('Fins')
            plt.show()
        else:
            print(f"No data for {class_type} to create hexbin or scatter plot.")

    # Swarm plot of conservation status across classes
    plt.figure(figsize=(14, 8))
    sns.swarmplot(data=final_merged_df, x='Class_Type', y='trophic_level', hue='conservation_status', palette='muted', dodge=True)
    plt.title('Trophic Level Distribution Across Classes by Conservation Status')
    plt.xlabel('Class Type')
    plt.ylabel('Trophic Level')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='Conservation Status', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

    # Clustermap of all numerical features
    numerical_features_df = final_merged_df.select_dtypes(include=['int64', 'float64'])
    correlation_matrix = numerical_features_df.corr()
    sns.clustermap(correlation_matrix, cmap='coolwarm', annot=True, fmt=".2f", linewidths=.5, figsize=(12,12))
    plt.suptitle('Clustermap of Correlation Matrix for All Numerical Features', y=1.02)
    plt.show()

    return final_merged_df

print("Function `beta_eda_and_cleaning` updated successfully with variance checks for hexbin plots.")
```

## Model Training and Evaluation Function: `beta_train_and_evaluate()`

### Subtask:
Create a function `beta_train_and_evaluate()` that prepares the data from `final_merged_df`, splits it into 80% training and 20% testing sets using `random_state=789`, and returns the split data.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier # Import Random Forest Classifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report # Import evaluation metrics

def beta_train_and_evaluate(df):
    """
    Prepares the data, splits it into training and testing sets,
    trains a Random Forest Classifier, evaluates it, and returns the split data.

    Args:
        df (pd.DataFrame): The input DataFrame (e.g., final_merged_df).

    Returns:
        tuple: X_train, X_test, y_train, y_test, trained_model after data splitting, preprocessing, and model training.
    """

    print("Starting data preparation and splitting...")

    # Define features (X) and target (y)
    # Target is 'Class_Type'
    X = df.drop(['Class_Type', 'animal_name'], axis=1) # Drop target and identifier column
    y = df['Class_Type']

    # Identify categorical and numerical features for preprocessing
    categorical_features = X.select_dtypes(include=['object']).columns
    numerical_features = X.select_dtypes(include=['int64', 'float64']).columns

    # Create a preprocessor to handle categorical features (OneHotEncoding)
    # and pass through numerical features
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', 'passthrough', numerical_features),
            ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_features)
        ])

    # Apply preprocessing
    # For simplicity, applying preprocessing before splitting here.
    # In a more rigorous ML pipeline, preprocessor would be fitted on training data only
    # and then transformed on both train and test to prevent data leakage.
    X_processed = preprocessor.fit_transform(X)

    # Convert X_processed back to DataFrame if it's a sparse matrix
    if hasattr(X_processed, 'toarray'):
        # Get feature names after one-hot encoding
        new_feature_names = list(numerical_features) + \
                            list(preprocessor.named_transformers_['cat'].get_feature_names_out(categorical_features))
        X_processed = pd.DataFrame(X_processed.toarray(), columns=new_feature_names)
    else:
        X_processed = pd.DataFrame(X_processed)

    # Split the data into training and testing sets
    X_train, X_test, y_train, y_test = train_test_split(
        X_processed, y, test_size=0.2, random_state=789, stratify=y
    )

    print("Data prepared and split successfully:")
    print(f"X_train shape: {X_train.shape}")
    print(f"X_test shape: {X_test.shape}")
    print(f"y_train shape: {y_train.shape}")
    print(f"y_test shape: {y_test.shape}")

    print("\nStarting Random Forest Model Training...")
    # Initialize the Random Forest Classifier with specified hyperparameters
    rf_model = RandomForestClassifier(
        n_estimators=80,
        max_depth=8,
        min_samples_split=6,
        random_state=789 # Use the same random state for reproducibility
    )

    # Train the model
    rf_model.fit(X_train, y_train)
    print("Random Forest Model trained successfully.")

    # Make predictions on the test set
    y_pred = rf_model.predict(X_test)

    print("\nModel Evaluation:")
    # Evaluate the model
    accuracy = accuracy_score(y_test, y_pred)
    precision = precision_score(y_test, y_pred, average='weighted', zero_division=0)
    recall = recall_score(y_test, y_pred, average='weighted', zero_division=0)
    f1 = f1_score(y_test, y_pred, average='weighted', zero_division=0)

    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"F1-Score: {f1:.4f}")

    print("\nClassification Report:")
    print(classification_report(y_test, y_pred, zero_division=0))

    return X_train, X_test, y_train, y_test, rf_model

print("Function `beta_train_and_evaluate` defined successfully, ready for use.")
```
